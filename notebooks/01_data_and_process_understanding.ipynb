## 2. Industrial Context

SWaT is a scaled-down, high-fidelity water-treatment testbed developed for research into the security and operation of cyber-physical systems. Its physical process consists of six interconnected treatment stages. The testbed combines physical equipment with industrial control and monitoring components, including Programmable Logic Controllers (PLCs), Human-Machine Interfaces (HMIs), a Supervisory Control and Data Acquisition (SCADA) workstation, and a historian that records process data for subsequent analysis [1].

The physical process and its control system must be considered together. Measurements describe observed process conditions, while actuator states and control logic help explain how equipment is operated. A measurement should therefore be interpreted in its process context rather than treated as abnormal solely because its value changes.

## 3. Water-Treatment Process Stages

The Kaggle dataset description presents SWaT as a multi-stage water-treatment process [1]. The following high-level summary provides process context; it does not imply that every stage or process variable has been identified in the local copy.

### 3.1 P1 — Raw Water Intake and Storage

Raw water enters the system and is held in a storage tank before moving to the pretreatment stage.

### 3.2 P2 — Pretreatment

The water undergoes pretreatment, including disinfection and pH balancing, before filtration.

### 3.3 P3 — Ultrafiltration

The water passes through an ultrafiltration (UF) unit as part of the treatment process.

### 3.4 P4 — Dechlorination

The process includes dechlorination before reverse osmosis.

### 3.5 P5 — Reverse Osmosis

The water enters a reverse-osmosis (RO) system, which produces two streams: RO permeate and RO reject.

### 3.6 P6 — Storage and Ultrafiltration Backwash

The process includes storage and ultrafiltration backwash. Details of the local testbed configuration are not established by this project notebook [1].

### 3.7 Process Interdependence

The stages form a connected process, so changes in one stage may affect conditions observed elsewhere. This motivates examining relationships among variables and their behaviour over time, rather than relying only on isolated measurements. The exact relationships that can be assessed in this project will depend on the variables and documentation available for the selected dataset.

## 4. Instrumentation and Control

The Kaggle dataset description identifies sensor and actuator measurements from the SWaT testbed. In general, sensors measure process conditions and actuators carry out physical actions such as regulating flow [1].

For anomaly detection, the relevant evidence may include unusual measurements, unexpected actuator states, inconsistent relationships between variables, or changes in temporal patterns. Such observations identify behaviour for investigation; they do not, by themselves, establish that a cyberattack, equipment failure, or other specific cause has occurred.

Variable names and data types are examined in the project's data-quality work. The Kaggle listing does not establish detailed units or process locations for every variable in the local copy.

## 5. Problem Definition

The project investigates whether patterns in multivariate time-series data can be used to detect observations or periods that deviate from expected operating behaviour. The evaluation will consider both individual anomalous observations and sustained deviations over time.

Where the selected dataset provides labels, those labels will be examined and used as an evaluation reference only after their definitions and limitations have been verified. A detected deviation will not automatically be interpreted as a confirmed attack or physical failure.

## 6. Study Scope

The planned study will compare nine anomaly-detection methods across three categories:

- Statistical and rule-based baselines.
- Traditional machine-learning methods.
- Deep-learning methods.

The comparison is intended to examine detection performance, false alarms, event-level detection behaviour, detection delay, and robustness under a consistent evaluation framework. The final method list, metric definitions, event-matching rules, and evaluation protocol must be documented before model evaluation; this notebook does not report model results.

## 7. Limitations and Interpretation

The Kaggle listing identifies the reuploaded dataset but does not establish the underlying SWaT collection release represented by the local files. The exact upstream release therefore remains unverified, and dataset properties should not be inferred beyond the source page and the checks recorded in this project [1].

Anomaly-detection results can identify deviations from learned or defined behaviour, but they cannot independently establish the physical cause or intent behind those deviations. Conclusions will be limited to what the inspected data, verified labels, and supporting process documentation can substantiate. Results from a testbed should also not be assumed to generalize automatically to every full-scale water-treatment facility.
